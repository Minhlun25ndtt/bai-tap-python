Bài 1.1.

(Dự đoán kết quả)  Với x = 23 và y = 6, hãy tính tay giá trị các biểu thức x / y, x // y, x % y, x ** 2, -x // y, -x % y, 2 ** 3 ** 2, (2 ** 3) ** 2; sau đó kiểm tra lại bằng máy.

```python
x = 23
y = 6

print("x / y =", x / y)           # Chia lấy số thực: 3.8333...
print("x // y =", x // y)         # Chia lấy phần nguyên: 3
print("x % y =", x % y)           # Chia lấy phần dư: 5
print("x ** 2 =", x ** 2)         # Lũy thừa: 529

print("-x // y =", -x // y)       # -23 // 6 = -4 (Làm tròn xuống số nguyên nhỏ hơn)
print("-x % y =", -x % y)         # -23 % 6 = 1 (Vì -23 = 6 * (-4) + 1)

print("2 ** 3 ** 2 =", 2 ** 3 ** 2)       # Lũy thừa tính từ phải qua trái: 2 ** (3 ** 2) = 2 ** 9 = 512
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)   # Có ngoặc tính trước: 8 ** 2 = 64
```

Bài 1.2.

(Dự đoán kết quả)  Cho s = "KY THUAT LAP TRINH". Dự đoán kết quả của len(s), s[0], s[-1], s[3:8], s[:2], s[::-1][:4], s[::3] và s[30:]. Kiểm tra lại bằng máy.

```python
s = "KY THUAT LAP TRINH"

print("len(s) =", len(s))               # Độ dài chuỗi: 18
print("s[0] =", s[0])                   # Ký tự đầu tiên: 'K'
print("s[-1] =", s[-1])                 # Ký tự cuối cùng: 'H'
print("s[3:8] =", s[3:8])               # Cắt từ index 3 đến 8: 'THUAT'
print("s[:2] =", s[:2])                 # Cắt từ đầu đến index 2: 'KY'
print("s[::-1][:4] =", s[::-1][:4])     # Đảo ngược chuỗi rồi lấy 4 ký tự đầu: 'HNIR'
print("s[::3] =", s[::3])               # Lấy các ký tự cách nhau 3 bước: 'K TAL I'
print("s[30:] =", s[30:])               # Index vượt quá độ dài chuỗi: in ra chuỗi rỗng ''
```

(Dự đoán kết quả)  Dự đoán đầu ra của bốn dòng dưới đây rồi chạy để kiểm tra. Chú ý các giá trị làm tròn và phép ép kiểu; giải thích vì sao round(2.5) và round(3.5) không cho kết quả “tăng một đơn vị” như nhau.

print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
print(int(-3.7), -3.7 // 1, int(3.99))
print(int("12") + int("3"), "12" + "3")
print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)

```python
print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
# Output: 2 4 0 1.23 (Làm tròn chữ số thập phân)

print(int(-3.7), -3.7 // 1, int(3.99))
# Output: -3 -4.0 3 (int() cắt bỏ phần thập phân, // làm tròn xuống)

print(int("12") + int("3"), "12" + "3")
# Output: 15 123 (Phép cộng số nguyên vs Nối chuỗi)

print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)
# Output: abababc 7.0 -2 ("ab" lặp 3 lần rồi nối "c"; 7.0 vì phép chia luôn ra float; 10 - (4*3) = -2)
```

Giải thích: Trong Python, hàm round() sử dụng quy tắc "Banker's rounding" (làm tròn đến số chẵn gần nhất nếu phần thập phân chính xác là 0.5). Do đó:

round(2.5) làm tròn về số chẵn gần nhất là 2.

round(3.5) làm tròn về số chẵn gần nhất là 4.

(Dự đoán kết quả)  Dự đoán chính xác những gì chương trình dưới đây in ra (chú ý khoảng trắng, dấu xuống dòng và định dạng số), rồi chạy kiểm tra.

a, b = 7, 2
print(a, b, sep="")
print(a / b, a // b, sep=" | ")
print(f"{a}+{b}={a + b}", end="; ")
print(f"{a / b:.1f}")
print("a" "b", 3 * "ab")
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")

```python
a, b = 7, 2
print(a, b, sep="")                      # Output: 72 (Không có khoảng trắng ở giữa do sep="")
print(a / b, a // b, sep=" | ")          # Output: 3.5 | 3 (Ngăn cách bởi " | ")
print(f"{a}+{b}={a + b}", end="; ")      # In ra "7+2=9; " và KHÔNG xuống dòng
print(f"{a / b:.1f}")                    # Output: 3.5 (In tiếp trên cùng dòng do lệnh print trên có end="; ", định dạng 1 chữ số thập phân)
print("a" "b", 3 * "ab")                 # Output: ab ababab (Tự động nối chuỗi "a" và "b", lặp "ab" 3 lần)
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")
# Output: 1,234.57 005 25.6% (Dấu phẩy phân cách hàng nghìn & 2 số thập phân, padding 3 chữ số 0, định dạng phần trăm)
```

(Đọc lỗi và sửa mã)  Đoạn mã dưới đây tính chỉ số khối cơ thể

can_nang = input("Cân nặng (kg): ")
chieu_cao = input("Chiều cao (m): ")
bmi = can_nang / chieu_cao ^ 2
print(f"BMI = {bmi:.1f}")

```python
can_nang = input("Cân nặng (kg): ")
chieu_cao = input("Chiều cao (m): ")
bmi = can_nang / chieu_cao ^ 2
print(f"BMI = {bmi:.1f}")
```

Hai lỗi sai trong đoạn mã gốc:

Hàm input() luôn trả về giá trị kiểu chuỗi (str), nhưng công thức tính BMI yêu cầu số thực. Nếu không ép kiểu sang float(), máy sẽ báo lỗi TypeError khi thực hiện phép chia.

Ký hiệu lũy thừa trong Python là **, trong khi mã gốc dùng ^ .

```python
# Ép kiểu dữ liệu đầu vào sang float (số thực)
can_nang = float(input("Cân nặng (kg): "))
chieu_cao = float(input("Chiều cao (m): "))

# Sửa lỗi phép lũy thừa từ ^ thành **
bmi = can_nang / (chieu_cao ** 2)

print(f"BMI = {bmi:.1f}")
```
