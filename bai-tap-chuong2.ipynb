## TH 2.2

Giải đầy đủ $ax^2+bx+c=0$ cho mọi hệ số: nếu $a=0$ thì là phương trình bậc nhất $bx+c=0$ (có thể vô nghiệm hoặc vô số nghiệm). Kiểm tra với $(0,2,-4)$, $(0,0,5)$, $(0,0,0)$, $(1,-5,6)$, $(2,4,2)$, $(1,1,1)$.

```python
# TODO: viết lời giải của bạn ở đây
import math
for a, b, c in [(0, 2, -4), (0, 0, 5), (0, 0, 0), (1, -5, 6), (2, 4, 2), (1, 1, 1)]:
    kq = f"({a},{b},{c}): "
    if a == 0 and b == 0:
        kq += "Vô số nghiệm" if c == 0 else "Vô nghiệm"
    elif a == 0:
        kq += f"Có một nghiệm x = {-c / b}"
    else:
        delta = b**2 - 4*a*c
        if delta > 0:
            x1 = (-b + math.sqrt(delta)) / (2*a)
            x2 = (-b - math.sqrt(delta)) / (2*a)
            kq += f"delta={delta} > 0, hai nghiệm {x1} và {x2}"
        elif delta == 0:
            kq += f"delta=0, nghiệm kép {-b / (2*a)}"
        else:
            kq += f"delta={delta} < 0, vô nghiệm thực"
    print(kq)
```

## TH 2.3

Viết chương trình tính lương tuần như Ví dụ \ref{vd:luong} (lương $100 000$ đồng/giờ, giờ vượt $40$ tính gấp $1{,}5$) nhưng *nhập lại* cho đến khi người dùng gõ một số *không âm*: chữ hoặc số âm đều phải báo lỗi và hỏi lại. Kiểm tra với dãy gõ `abc`, `-2`, `45`.

```python
# TODO: viết lời giải của bạn ở đây
while True:
    try:
        gio = float(input("Số giờ làm: "))
        if gio >= 0:
            break
        print("Lỗi: số không được âm. Hãy nhập lại.")
    except ValueError:
        print("Lỗi: phải nhập một số. Hãy nhập lại.")

luong = 100000 * (min(gio, 40) + 1.5 * max(gio - 40, 0))
print(f"Lương tuần của bạn là {luong:,.0f} đồng")
```

## TH 2.7

Dãy Collatz: từ $n$, nếu $n$ chẵn thì $n\leftarrow n/2$, nếu lẻ thì $n\leftarrow3n+1$, lặp đến khi $n=1$. Với $n=27$, đếm số bước và tìm giá trị lớn nhất mà dãy đạt được (đáp án biết trước: $111$ bước, giá trị lớn nhất $9232$).

```python
# TODO: viết lời giải của bạn ở đây
n = 27
so_buoc = 0
gia_tri_max = n

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    so_buoc += 1
    if n > gia_tri_max:
        gia_tri_max = n

print(f"Số bước: {so_buoc}")
print(f"Giá trị lớn nhất: {gia_tri_max}")
```

## TH 2.10 ∗

Mở rộng Ví dụ \ref{vd:montepi}: ước lượng $\pi$ với $N=10^2,\ 10^4,\ 10^6$ điểm (dùng `random.seed(2026)` một lần ở đầu). In ước lượng và sai số tuyệt đối; kiểm tra bằng `assert` rằng sai số nhỏ hơn $6/\sqrt N$. Nhận xét: gấp $100$ lần số điểm thì sai số giảm khoảng mấy lần?

```python
# TODO: viết lời giải của bạn ở đây
import random
import math

random.seed(2026)

for N in [10**2, 10**4, 10**6]:
    trong = sum(random.random()**2 + random.random()**2 <= 1 for _ in range(N))
    pi_uoc = 4 * trong / N
    sai_so = abs(pi_uoc - math.pi)
    print(f"N = {N:>8}: pi ≈ {pi_uoc:.6f}, sai số = {sai_so:.6f}")
    assert sai_so < 6 / math.sqrt(N)
```

## TH 2.11 ∗

Phương trình $f(x)=x^3-x-2=0$ có đúng một nghiệm trong $[1,2]$ vì $f(1)=-2<0<4=f(2)$ và $f$ tăng trên đoạn này. Cài đặt *phương pháp chia đôi*: giữ đoạn $[a,b]$ mà $f(a)f(b)<0$, thay bằng nửa đoạn chứa nghiệm cho tới khi $b-a<10^{-10}$. In nghiệm, số bước, và so sánh với kết quả lý thuyết: số bước cần thiết là $\lceil\log_2(1/10^{-10})\rceil$.

```python
# TODO: viết lời giải của bạn ở đây
import math

def f(x):
    return x**3 - x - 2

a, b = 1.0, 2.0
so_buoc = 0

while b - a >= 1e-10:
    m = (a + b) / 2
    if f(a) * f(m) <= 0:
        b = m
    else:
        a = m
    so_buoc += 1

print(f"Nghiệm x ≈ {(a + b) / 2:.10f}")
print(f"Số bước thực tế: {so_buoc}, lý thuyết: {math.ceil(math.log2(1e10))}")
```
