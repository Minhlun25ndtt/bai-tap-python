## TH 3.2

Điểm của $8$ giám khảo là `[7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]`. Để giảm ảnh hưởng của giá trị bất thường, bỏ đi $n=2$ điểm cao nhất và $n=2$ điểm thấp nhất rồi tính trung bình các điểm còn lại (*trung bình cắt đầu--cuối*). So sánh với trung bình thông thường; *không* sửa danh sách gốc.

```python
# TODO: viết lời giải của bạn ở đây
diem = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]
n = 2

con_lai = sorted(diem)[n:len(diem) - n]

print(f"Trung bình thông thường: {sum(diem) / len(diem):.3f}")
print(f"Trung bình cắt đầu-cuối: {sum(con_lai) / len(con_lai):.3f}")
```

## TH 3.4

Số giờ ôn tập $x$ và điểm thi $y$ của $5$ sinh viên là `[(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]`. Tìm đường hồi quy tuyến tính $y=a+bx$ theo công thức $b=\dfrac{\sum(x_i-\bar x)(y_i-\bar y)}{\sum(x_i-\bar x)^2}$, $a=\bar y-b\bar x$; dự đoán điểm của sinh viên ôn $6$ giờ; kiểm tra bằng `statistics.linear_regression`.

```python
# TODO: viết lời giải của bạn ở đây
import statistics

du_lieu = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]
x = [p[0] for p in du_lieu]
y = [p[1] for p in du_lieu]

x_tb = sum(x) / len(x)
y_tb = sum(y) / len(y)

Sxy = sum((xi - x_tb) * (yi - y_tb) for xi, yi in du_lieu)
Sxx = sum((xi - x_tb) ** 2 for xi in x)

b = Sxy / Sxx
a = y_tb - b * x_tb

print(f"x̄ = {x_tb}, ȳ = {y_tb}, Sxy = {Sxy:.4g}, Sxx = {Sxx:.4g}")
print(f"b = {b:.4f}, a = {a:.4f}")
print(f"Dự đoán y(6) = {a + b * 6:.4f}")

slope, intercept = statistics.linear_regression(x, y)
print(f"statistics.linear_regression: b = {slope:.4f}, a = {intercept:.4f}")
assert abs(a - intercept) < 1e-9 and abs(b - slope) < 1e-9
print("Khớp với statistics.linear_regression")
```

## TH 3.6

Cho danh sách các cặp `(tên, điểm)`: `[("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]`. Quy đổi sang điểm chữ ($\ge8{,}5$: A; $\ge7$: B; $\ge5{,}5$: C; $\ge4$: D; còn lại F) rồi gom tên sinh viên theo điểm chữ bằng từ điển (dùng `defaultdict(list)`). In các nhóm theo thứ tự chữ cái.

```python
# TODO: viết lời giải của bạn ở đây
from collections import defaultdict

sinh_vien = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0),
             ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]

def diem_chu(d):
    if d >= 8.5:
        return "A"
    elif d >= 7:
        return "B"
    elif d >= 5.5:
        return "C"
    elif d >= 4:
        return "D"
    else:
        return "F"

nhom = defaultdict(list)
for ten, diem in sinh_vien:
    nhom[diem_chu(diem)].append(ten)

for chu in sorted(nhom):
    print(f"{chu}: {', '.join(nhom[chu])}")
```

## TH 3.8

Doanh thu theo sản phẩm của hai cửa hàng: "ao": 120, "quan": 80, "non": 30 và "ao": 90, "giay": 150, "non": 20. Hợp nhất thành tổng doanh thu theo sản phẩm, in theo thứ tự giảm dần; tìm các sản phẩm bán ở cả hai cửa hàng và các sản phẩm chỉ bán ở một cửa hàng (dùng phép toán tập hợp trên khoá).

```python
# TODO: viết lời giải của bạn ở đây
ch1 = {"ao": 120, "quan": 80, "non": 30}
ch2 = {"ao": 90, "giay": 150, "non": 20}

tong = ch1.copy()
for sp, dt in ch2.items():
    tong[sp] = tong.get(sp, 0) + dt

print("Tổng doanh thu (giảm dần):")
for sp, dt in sorted(tong.items(), key=lambda kv: kv[1], reverse=True):
    print(f"  {sp}: {dt}")

print("Bán ở cả hai cửa hàng:", sorted(ch1.keys() & ch2.keys()))
print("Chỉ bán ở một cửa hàng:", sorted(ch1.keys() ^ ch2.keys()))
```

## TH 3.10
Dùng sàng Eratosthenes (cài đặt bằng set, theo Mệnh đề 3.2) tìm các số nguyên tố không vượt quá 1000: in số lượng, tổng, và số cặp số nguyên tố sinh đôi (p,p+2)cùng ≤1000. Kiểm tra: có 168 số nguyên tố, tổng bằng 76127

```python
# TODO: viết lời giải của bạn ở đây
N = 1000

nt = set(range(2, N + 1))
for p in range(2, int(N ** 0.5) + 1):
    if p in nt:
        nt -= set(range(p * p, N + 1, p))

sinh_doi = [(p, p + 2) for p in sorted(nt) if p + 2 in nt]

print(f"Số lượng số nguyên tố ≤ {N}: {len(nt)}")
print(f"Tổng: {sum(nt)}")
print(f"Số cặp sinh đôi: {len(sinh_doi)}, cặp đầu: {sinh_doi[0]}")

assert len(nt) == 168 and sum(nt) == 76127
```
