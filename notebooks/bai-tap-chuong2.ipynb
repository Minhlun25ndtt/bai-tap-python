## TH 2.2

Giải đầy đủ $ax^2+bx+c=0$ cho mọi hệ số: nếu $a=0$ thì là phương trình bậc nhất $bx+c=0$ (có thể vô nghiệm hoặc vô số nghiệm). Kiểm tra với $(0,2,-4)$, $(0,0,5)$, $(0,0,0)$, $(1,-5,6)$, $(2,4,2)$, $(1,1,1)$.

```python
# TODO: viết lời giải của bạn ở đây
import math
test = ((0, 2, -4), (0, 0, 5), (0, 0, 0), (1, -5, 6), (2, 4, 2), (1, 1, 1))
for a, b, c in test:
    if a == 0:
        if b == 0:
            if c == 0:
                print(f"({a},{b},{c}): Vô số nghiệm")
            else:
                print(f"({a},{b},{c}): Vô nghiệm")
        else:
            x = -c / b
            print(f"({a},{b},{c}): Có một nghiệm x = {x}")
    else:
        delta = b ** 2 - 4 * a * c
        if delta > 0:
            x1 = (-b + math.sqrt(delta)) / (2 * a)
            x2 = (-b - math.sqrt(delta)) / (2 * a)
            print(f"({a},{b},{c}): delta={delta} > 0, hai nghiệm {x1} và {x2}")
        elif delta == 0:
            x = -b / (2 * a)
            print(f"({a},{b},{c}): delta=0, nghiệm kép {x}")
        else:
            print(f"({a},{b},{c}): delta={delta} < 0, vô nghiệm thực")
```

## TH 2.3

Viết chương trình tính lương tuần như Ví dụ \ref{vd:luong} (lương $100 000$ đồng/giờ, giờ vượt $40$ tính gấp $1{,}5$) nhưng *nhập lại* cho đến khi người dùng gõ một số *không âm*: chữ hoặc số âm đều phải báo lỗi và hỏi lại. Kiểm tra với dãy gõ `abc`, `-2`, `45`.

```python
# TODO: viết lời giải của bạn ở đây
def nhap_so_khong_am(thong_bao):
    """Hỏi lại cho đến khi người dùng nhập một số không âm."""
    while True:
        try:
            so = float(input(thong_bao))
        except ValueError:
            print("Lỗi: bạn phải nhập một số, không được nhập chữ. Hãy nhập lại.")
            continue
        if so < 0:
            print("Lỗi: số không được âm. Hãy nhập lại.")
            continue
        return so

luong_gio = 100000
gio_chuan = 40

gio_lam = nhap_so_khong_am("Số giờ làm: ")

if gio_lam <= gio_chuan:
    luong = gio_lam * luong_gio
else:
    luong = gio_chuan * luong_gio + (gio_lam - gio_chuan) * luong_gio * 1.5

print(f"Lương tuần của bạn là {luong:,.0f} đồng")
```

## TH 2.4

Nhập liên tục các số thực (bỏ qua và báo giá trị không phải số) cho đến khi gõ `xong`. In số giá trị hợp lệ, tổng, trung bình, giá trị lớn nhất và nhỏ nhất; nếu chưa nhập số nào thì báo ``chưa có dữ liệu''. Kiểm tra với dãy gõ: `7.5, 8, abc, 6.5, 9, xong`.

```python
# TODO: viết lời giải của bạn ở đây
so_hop_le = []

while True:
    s = input("Nhập một số (gõ 'xong' để kết thúc): ").strip()

    if s.lower() == "xong":
        break

    try:
        so = float(s)
    except ValueError:
        print(f"Bỏ qua: '{s}' không phải là số.")
        continue

    so_hop_le.append(so)

if len(so_hop_le) == 0:
    print("Chưa có dữ liệu")
else:
    tong = sum(so_hop_le)
    print(f"Số giá trị hợp lệ: {len(so_hop_le)}")
    print(f"Tổng: {tong:g}")
    print(f"Trung bình: {tong / len(so_hop_le):g}")
    print(f"Lớn nhất: {max(so_hop_le):g}")
    print(f"Nhỏ nhất: {min(so_hop_le):g}")
```

## TH 2.7

Dãy Collatz: từ $n$, nếu $n$ chẵn thì $n\leftarrow n/2$, nếu lẻ thì $n\leftarrow3n+1$, lặp đến khi $n=1$. Với $n=27$, đếm số bước và tìm giá trị lớn nhất mà dãy đạt được (đáp án biết trước: $111$ bước, giá trị lớn nhất $9232$).

```python
# TODO: viết lời giải của bạn ở đây
n = 27
so_buoc = 0
gia_tri_max = n

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    so_buoc += 1
    if n > gia_tri_max:
        gia_tri_max = n

print(f"Số bước: {so_buoc}")
print(f"Giá trị lớn nhất: {gia_tri_max}")
```

## TH 2.10 ∗

Mở rộng Ví dụ \ref{vd:montepi}: ước lượng $\pi$ với $N=10^2,\ 10^4,\ 10^6$ điểm (dùng `random.seed(2026)` một lần ở đầu). In ước lượng và sai số tuyệt đối; kiểm tra bằng `assert` rằng sai số nhỏ hơn $6/\sqrt N$. Nhận xét: gấp $100$ lần số điểm thì sai số giảm khoảng mấy lần?

```python
# TODO: viết lời giải của bạn ở đây
import random
import math

random.seed(2026)

for N in [10**2, 10**4, 10**6]:
    trong_tron = 0
    for _ in range(N):
        x = random.random()
        y = random.random()
        if x*x + y*y <= 1:
            trong_tron += 1
    pi_uoc_luong = 4 * trong_tron / N
    sai_so = abs(pi_uoc_luong - math.pi)
    nguong = 6 / math.sqrt(N)
    print(f"N = {N:>8}: pi ≈ {pi_uoc_luong:.6f}, sai số = {sai_so:.6f}, ngưỡng 6/√N = {nguong:.6f}")
    assert sai_so < nguong, f"Sai số vượt ngưỡng với N = {N}"

print("Tất cả assert đều đạt.")
```

## TH 2.11 ∗

Phương trình $f(x)=x^3-x-2=0$ có đúng một nghiệm trong $[1,2]$ vì $f(1)=-2<0<4=f(2)$ và $f$ tăng trên đoạn này. Cài đặt *phương pháp chia đôi*: giữ đoạn $[a,b]$ mà $f(a)f(b)<0$, thay bằng nửa đoạn chứa nghiệm cho tới khi $b-a<10^{-10}$. In nghiệm, số bước, và so sánh với kết quả lý thuyết: số bước cần thiết là $\lceil\log_2(1/10^{-10})\rceil$.

```python
# TODO: viết lời giải của bạn ở đây
import math

def f(x):
    return x**3 - x - 2

a, b = 1.0, 2.0
dung_sai = 1e-10
so_buoc = 0

assert f(a) * f(b) < 0, "Đoạn [a, b] không thỏa f(a)f(b) < 0"

while b - a >= dung_sai:
    m = (a + b) / 2
    if f(a) * f(m) <= 0:
        b = m
    else:
        a = m
    so_buoc += 1

nghiem = (a + b) / 2
ly_thuyet = math.ceil(math.log2(1 / 1e-10))

print(f"Nghiệm x ≈ {nghiem:.10f}")
print(f"f(x) ≈ {f(nghiem):.2e}")
print(f"Số bước thực tế: {so_buoc}")
print(f"Số bước lý thuyết: ceil(log2(1/1e-10)) = {ly_thuyet}")
```
